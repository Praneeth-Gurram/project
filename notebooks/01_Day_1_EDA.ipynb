## 1. Import Required Libraries
We will begin by importing the foundational tools used by Data Analysts in Python:
- **Pandas**: Used for data manipulation and analysis.
- **NumPy**: Used for numerical computing.
- **Matplotlib**: Used for creating static, interactive, and animated visualizations.

```python
# Import essential data analysis libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# (Optional) Set Pandas options to display all columns for easier viewing
pd.set_option('display.max_columns', None)
```

## 2. Load the Dataset
Next, we load the `smart_logistics.csv` dataset. By storing data in a structured folder (like `data/raw/`), we ensure our project remains organized.

```python
# Define the path to the dataset
# Assuming this notebook is inside the 'notebooks' directory and the dataset is in 'data/raw'
file_path = '../data/raw/smart_logistics.csv'

# Load the dataset into a Pandas DataFrame
df = pd.read_csv(file_path)

print("Dataset successfully loaded!")
```

## 3. Initial Data Inspection
Now that the dataset is loaded, we will use standard Pandas functions to peek at the data and gather structural information.

### Display `head()`
The `head()` function displays the first 5 rows of the DataFrame. 
**Why we use it:** It gives us a quick visual confirmation that the data loaded correctly and lets us see what the raw data looks like.

```python
# View the first 5 rows of the DataFrame
df.head()
```

### Display `tail()`
The `tail()` function displays the last 5 rows of the DataFrame. 
**Why we use it:** It helps us check if there is any corrupted data or unexpected summary rows appended at the very end of our file.

```python
# View the last 5 rows of the DataFrame
df.tail()
```

### Display `shape`
The `shape` attribute returns a tuple representing `(number of rows, number of columns)`.
**Why we use it:** It immediately tells us the size of our dataset, giving us an idea of the scale of the analysis.

```python
# Check the dimensions of the dataset (rows, columns)
df.shape
```

### Display `columns`
The `columns` attribute lists all column names (features) in the dataset.
**Why we use it:** It provides a handy reference for all the variables we have available to analyze.

```python
# List all column names in the dataset
df.columns
```

### Display `info()`
The `info()` function provides a concise summary of the DataFrame, including data types, non-null counts, and memory usage.
**Why we use it:** It is critical for identifying columns that have missing values (nulls) and verifying that data types are correctly assigned (e.g., making sure dates are not loaded as plain strings).

```python
# Get a concise summary of the DataFrame (data types, missing values, memory usage)
df.info()
```

### Display `describe()`
The `describe()` function generates descriptive statistics for all numerical columns (count, mean, std, min, max, and quartiles).
**Why we use it:** It helps us instantly understand the distribution of our numerical variables and spot any obvious outliers (like negative costs or extreme weights).

```python
# Generate descriptive statistics for numerical columns
df.describe()
```
